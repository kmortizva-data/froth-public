## 1) Preparar todo
Esta celda carga el proyecto, la tabla de papers y sus vectores. Ejecútala una vez (Shift+Enter).

```python
# Tell Python where the project lives (so we can import 'litmap').
import sys
sys.path.insert(0, r"C:\Users\Bruce Wayne\Documents\Master Thesis AI")

import pandas as pd
from litmap import config, embed

# Load the papers table and the precomputed vector matrix.
df = pd.read_parquet(config.DATA_PROCESSED / "papers.parquet")
V = embed.load_embeddings()
print(f"Loaded {len(df)} papers and their vectors {V.shape}.")
```

## 2) Escribe tu consulta 👇
Cambia el texto entre comillas (la variable `my_query`) por lo que quieras buscar y ejecuta la celda (Shift+Enter).
El resultado sale como tabla: **score** (0 a 1, más alto = más parecido), título, año y citas.

```python
my_query = "effect of particle size on lithium mica flotation recovery"

embed.most_similar(my_query, df, V, k=5)
```
